```python
import h5py as h5
import numpy as np
import pandas as pd
from tqdm import tqdm
fname = "/mnt/ceph/users/spandey/xdesi/des/clusters/halo_lightcone_base_c000_ph000_full.h5"

columns = (
    "M_interp",
    "X_interp",
    "Y_interp",
    "Z_interp",
    "redshift_interp",
)

block_size = 10_000_000
selected = []

with h5.File(fname, "r") as f:
    halos = f["halos"]
    fields = halos.fields(columns)

    for start in tqdm(range(0, len(halos), block_size)):
        block = fields[start : start + block_size]
        selected.append(block[block["M_interp"] > 1.0e13])

halos_selected = np.concatenate(selected)


```
